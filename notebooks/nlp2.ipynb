```python
import json, glob
def load(m): return {json.load(open(f))['pair_id']: json.load(open(f)) for f in glob.glob(f'/content/drive/MyDrive/sp-estimation/results/{m}/pairs/pair_*.json')}
c, s, b = load('combined'), load('sbert'), load('bert')
print(len(c), len(s), len(b))
print('split sizes differ:', [p for p in b if (b[p]['n_source_dev'], b[p]['n_target_test']) != (c[p]['n_source_dev'], c[p]['n_target_test'])])
print('backends:', {r['backend'] for r in list(c.values()) + list(s.values()) + list(b.values())})
```
